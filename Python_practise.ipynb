```python
images = [
    {"file": "img1.jpg", "boxes": [{"cls": "scratch"}, {"cls": "dent"}, {"cls": "scratch"}]},
    {"file": "img2.jpg", "boxes": [{"cls": "dent"}]},
    {"file": "img3.jpg", "boxes": []},
    {"file": "img4.jpg", "boxes": [{"cls": "chipping"}, {"cls": "scratch"}]},
]
```

```python
#Expected: {'scratch': 3, 'dent': 2, 'chipping': 1}.
#Hint: the outer loop goes over images, the inner loop goes over that image’s boxes, and counts[cls] = counts.get(cls, 0) + 1 does the counting.
```

```python
counts = {}
for i in images:
  for j in i["boxes"]:
    cls = j['cls']
    counts[cls] = counts.get(cls,0) + 1
print(counts)
```

```python
for i in images:
  total = len(i['boxes'])

  if total == 0:
    print(i['file'], "has no boxes")

  print(i['file'], ":", total)
```
